```python
#10.Palindrome:
# Import the 'string' module (though not strictly necessary for this specific implementation, it's good practice if more string ops were needed)
import string

# Prompt the user to enter a string
text = input("Enter string: ")

# Initialize an empty string to store the cleaned version of the input
clean = ""

# Iterate through each character in the input string
for ch in text:
    # Check if the character is alphanumeric (a-z, A-Z, 0-9)
    if ch.isalnum():
        # If it is, convert it to lowercase and append it to the 'clean' string
        clean += ch.lower()

# Compare the cleaned string with its reverse
# [::-1] creates a reversed copy of the string
if clean == clean[::-1]:
    # If they are the same, the original string (after cleaning) is a palindrome
    print("Palindrome")
else:
    # Otherwise, it is not a palindrome
    print("Not Palindrome")
```
