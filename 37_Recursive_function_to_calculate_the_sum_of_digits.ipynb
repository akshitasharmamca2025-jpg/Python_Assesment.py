```python

import time

# 37.Recursive function to calculate the sum of digits
def sum_digits_recursive(n):
    """
    Calculates the sum of digits of a number using a recursive approach.
    Base case: if n is 0, sum is 0.
    Recursive step: adds the last digit (n % 10) to the sum of the remaining digits (n // 10).
    """
    if n == 0:
        return 0
    else:
        return (n % 10) + sum_digits_recursive(n // 10)

# Iterative function to calculate the sum of digits
def sum_digits_iterative(n):
    """
    Calculates the sum of digits of a number using an iterative approach.
    It repeatedly adds the last digit to a running total and removes the last digit
    until the number becomes 0.
    """
    total_sum = 0
    while n > 0:
        total_sum += n % 10  # Add the last digit
        n //= 10             # Remove the last digit
    return total_sum

# --- Comparison and Output ---

number_to_test = 123456789

print(f"Number: {number_to_test}")

# Test recursive approach
start_time = time.time()
recursive_result = sum_digits_recursive(number_to_test)
end_time = time.time()
print(f"Recursive sum of digits: {recursive_result} (Time taken: {end_time - start_time:.6f} seconds)")

# Test iterative approach
start_time = time.time()
iterative_result = sum_digits_iterative(number_to_test)
end_time = time.time()
print(f"Iterative sum of digits: {iterative_result} (Time taken: {end_time - start_time:.6f} seconds)")

# You can test with another number
# number_to_test_2 = 9876543210
# print(f"\nNumber: {number_to_test_2}")
# print(f"Recursive sum of digits: {sum_digits_recursive(number_to_test_2)}")
# print(f"Iterative sum of digits: {sum_digits_iterative(number_to_test_2)}")
```
