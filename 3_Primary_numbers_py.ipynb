```python
# =========================
# Q3 Prime Numbers
# =========================

# Taking input from user
n = int(input("Enter N: "))

# Empty list to store primes
primes = []

num = 2

# Loop until N prime numbers are found
while len(primes) < n:

    is_prime = True

    # Check prime condition
    for i in range(2, num):

        if num % i == 0:
            is_prime = False
            break

    # Add prime number to list
    if is_prime:
        primes.append(num)

    num += 1

# Display results
print("Prime Numbers:", primes)
print("Sum:", sum(primes))
print("Average:", sum(primes) / len(primes))
```
