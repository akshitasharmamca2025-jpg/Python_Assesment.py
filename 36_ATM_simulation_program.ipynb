```python
# Q36. ATM Simulation Program

def atm_simulation():
    """
    Simulates an ATM with PIN validation and account locking functionality.
    """
    correct_pin = "1234"
    attempts = 0
    max_attempts = 3
    account_locked = False

    print("Welcome to the ATM!")

    while attempts < max_attempts and not account_locked:
        # Prompt the user to enter their PIN
        user_pin = input(f"Enter your 4-digit PIN (Attempt {attempts + 1}/{max_attempts}): ")

        # Check if the entered PIN is correct
        if user_pin == correct_pin:
            print("PIN accepted! You can now access your account.")
            # In a real ATM, more options (e.g., withdraw, deposit, check balance) would follow here.
            break # Exit the loop if PIN is correct
        else:
            attempts += 1 # Increment attempts counter for incorrect PIN
            print("Incorrect PIN.")

    # After the loop, check if the account was locked
    if attempts == max_attempts and user_pin != correct_pin:
        account_locked = True # Ensure account_locked is true if max attempts reached with wrong PIN

    if account_locked:
        print("Too many incorrect attempts. Your account has been locked. Please contact your bank.")
    elif user_pin == correct_pin: # This condition catches the case where the loop broke due to correct PIN
        print("Thank you for using the ATM.")

# Run the ATM simulation
atm_simulation()
```
