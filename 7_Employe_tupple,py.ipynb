```python
# =========================
# Q7 Employee Tuple
# =========================

# Employee data using tuples
employees = [
    (101, "Aman", 30000),
    (102, "Riya", 45000),
    (103, "Karan", 25000)
]

# Calculate total salary
total = 0

for emp in employees:
    total += emp[2]

# Calculate average salary
avg = total / len(employees)

print("Average Salary:", avg)

# Display employees above average salary
print("Employees Above Average Salary:")

for emp in employees:

    if emp[2] > avg:
        print(emp)
```
