```python
# =========================
# Q17 Armstrong Numbers
# =========================

# Prompt the user to enter a number
number_str = input("Enter a number: ")

# Check if the input is a valid positive integer
if not number_str.isdigit():
    print("Invalid input. Please enter a positive integer.")
else:
    number = int(number_str)

    # An Armstrong number is a number that is equal to the sum of its own digits
    # each raised to the power of the number of digits.
    # Example: 153 = 1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153

    # Calculate the number of digits in the given number
    num_digits = len(number_str)

    # Initialize a variable to store the sum of the powers of the digits
    sum_of_powers = 0

    # Make a copy of the original number to manipulate in the loop
    temp_number = number

    # Loop through each digit of the number
    while temp_number > 0:
        # Get the last digit of the number
        digit = temp_number % 10

        # Add the digit raised to the power of num_digits to the sum
        sum_of_powers += digit ** num_digits

        # Remove the last digit from the number
        temp_number //= 10

    # Check if the sum of powers is equal to the original number
    if number == sum_of_powers:
        print(f"{number} is an Armstrong number.")
    else:
        print(f"{number} is not an Armstrong number.")
```
