```python
# Q39. Read a CSV file containing student attendance records and display students with attendance below 75%

import pandas as pd
import os

# 1. Create a dummy CSV file for demonstration purposes
# In a real scenario, you would already have an 'attendance.csv' file.
csv_data = """
StudentID,Name,ClassesAttended,TotalClasses
1,Alice,45,60
2,Bob,58,60
3,Charlie,30,45
4,David,50,60
5,Eve,40,50
6,Frank,25,35
7,Grace,55,60
"""

file_name = 'attendance.csv'

# Write the data to a CSV file
with open(file_name, 'w') as f:
    f.write(csv_data)

print(f"Created '{file_name}' with sample student attendance data.")

# 2. Read the attendance data from the CSV file into a Pandas DataFrame
try:
    df_attendance = pd.read_csv(file_name)
    print("\nAttendance Data Loaded:")
    print(df_attendance)
except FileNotFoundError:
    print(f"Error: '{file_name}' not found. Please ensure the file exists.")
    exit()
except Exception as e:
    print(f"An error occurred while reading the CSV: {e}")
    exit()

# 3. Calculate attendance percentage for each student
# The formula for percentage is (ClassesAttended / TotalClasses) * 100
df_attendance['AttendancePercentage'] = (df_attendance['ClassesAttended'] / df_attendance['TotalClasses']) * 100

print("\nAttendance Data with Percentage:")
print(df_attendance)

# 4. Identify students with attendance below 75%
threshold = 75
students_below_threshold = df_attendance[df_attendance['AttendancePercentage'] < threshold]

print(f"\nStudents with Attendance Below {threshold}%:")
if not students_below_threshold.empty:
    # Display only the relevant columns for these students
    print(students_below_threshold[['Name', 'AttendancePercentage']])
else:
    print(f"No students found with attendance below {threshold}%")

# Optional: Clean up the dummy CSV file after use
# os.remove(file_name)
# print(f"\nRemoved '{file_name}'.")
```
