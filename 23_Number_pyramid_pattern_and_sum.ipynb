```python
# =========================
# Q23 Number Pyramid Pattern and Sum
# =========================

# Prompt the user to enter the number of rows for the pyramid
try:
    rows = int(input("Enter the number of rows for the number pyramid: "))
except ValueError:
    print("Invalid input. Please enter an integer.")
else:
    if rows <= 0:
        print("Please enter a positive integer for the number of rows.")
    else:
        total_sum = 0
        print("\nNumber Pyramid Pattern:")

        # Outer loop for the number of rows
        for i in range(1, rows + 1):
            # Inner loop to print leading spaces for alignment
            for j in range(rows - i):
                print(" ", end="")

            # Inner loop to print numbers in increasing order
            for k in range(1, i + 1):
                print(k, end=" ")
                total_sum += k

            # Inner loop to print numbers in decreasing order (excluding the middle number)
            for l in range(i - 1, 0, -1):
                print(l, end=" ")
                total_sum += l

            print() # Move to the next line after each row

        print(f"\nSum of all numbers printed in the pyramid: {total_sum}")
```
