```python
import numpy as np

# =========================
# Q13 NumPy Matrix Operations
# =========================

# 1. Create a 5x5 matrix with random integers between 1 and 10
# np.random.randint(low, high, size) generates random integers from low (inclusive) to high (exclusive).
matrix = np.random.randint(1, 11, size=(5, 5))
print("Original 5x5 Matrix:\n", matrix)
print("\n" * 2)

# 2. Calculate row-wise sum
# axis=1 specifies that the sum should be calculated across rows (for each column).
row_sum = np.sum(matrix, axis=1)
print("Row-wise Sum:", row_sum)
print("\n" * 2)

# 3. Calculate column-wise sum
# axis=0 specifies that the sum should be calculated across columns (for each row).
column_sum = np.sum(matrix, axis=0)
print("Column-wise Sum:", column_sum)
print("\n" * 2)

# 4. Calculate the transpose of the matrix
# The .T attribute or np.transpose() function can be used to get the transpose.
matrix_transpose = matrix.T
print("Transpose of the Matrix:\n", matrix_transpose)
print("\n" * 2)

# 5. Calculate the determinant of the matrix
# np.linalg.det() computes the determinant of an array.
determinant = np.linalg.det(matrix)
print("Determinant of the Matrix:", determinant)
```
