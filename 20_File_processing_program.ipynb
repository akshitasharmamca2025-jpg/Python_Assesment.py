```python
# =========================
# Q20 File Processing Program
# =========================

def process_file(filename):
    """
    Processes a text file to count lines, words, and characters.
    Handles FileNotFoundError gracefully.
    """
    line_count = 0
    word_count = 0
    char_count = 0

    try:
        with open(filename, 'r') as file:
            for line in file:
                line_count += 1
                char_count += len(line)  # Count all characters including newline
                words = line.split()
                word_count += len(words)

        print(f"--- Analysis for '{filename}' ---")
        print(f"Total Lines: {line_count}")
        print(f"Total Words: {word_count}")
        print(f"Total Characters: {char_count}")

    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found. Please ensure the file exists and the path is correct.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Example Usage:
# 1. Create a dummy text file for testing
with open('sample.txt', 'w') as f:
    f.write('This is the first line.\n')
    f.write('This is the second line with more words.\n')
    f.write('And a third line.')

print("Created 'sample.txt' for testing.")

# Test with an existing file
process_file('sample.txt')

# Test with a non-existent file
print("\n")
process_file('non_existent_file.txt')
```
