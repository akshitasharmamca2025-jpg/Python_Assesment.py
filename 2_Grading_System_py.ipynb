```python
# =========================
# Q2 Grading System
# =========================

# Empty list to store marks
marks = []

# Taking marks input
for i in range(5):
    m = int(input(f"Enter marks of subject {i+1}: "))
    marks.append(m)

# Calculating percentage
percentage = sum(marks) / 5

# Assigning grade using conditions
if percentage >= 90:
    grade = "A"

elif percentage >= 75:
    grade = "B"

elif percentage >= 60:
    grade = "C"

else:
    grade = "D"

# Displaying result
print("Percentage:", percentage)
print("Grade:", grade)

# Scholarship condition
if percentage >= 85:
    print("Scholarship Qualified")

else:
    print("Not Qualified")
```
