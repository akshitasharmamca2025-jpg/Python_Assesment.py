```python
# Q28. Menu-driven mathematical utility program

import math

def calculate_factorial(n):
    """Calculates the factorial of a non-negative integer."""
    if not isinstance(n, int) or n < 0:
        return "Input must be a non-negative integer."
    return math.factorial(n)

def is_prime(n):
    """Checks if a number is prime."""
    if not isinstance(n, int):
        return "Input must be an integer."
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def is_armstrong(n):
    """Checks if a number is an Armstrong number."""
    if not isinstance(n, int) or n < 0:
        return "Input must be a non-negative integer."
    if n == 0:
        return True

    s = str(n)
    num_digits = len(s)
    sum_of_powers = 0
    for digit_char in s:
        sum_of_powers += int(digit_char) ** num_digits
    return n == sum_of_powers

def main_menu():
    """Displays the main menu and handles user choices."""
    while True:
        print("\n--- Mathematical Utility Program ---")
        print("1. Calculate Factorial")
        print("2. Check Prime Number")
        print("3. Check Armstrong Number")
        print("4. Exit")
        print("------------------------------------")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            try:
                num = int(input("Enter a non-negative integer for factorial: "))
                result = calculate_factorial(num)
                print(f"The factorial of {num} is: {result}")
            except ValueError:
                print("Invalid input. Please enter an integer.")

        elif choice == '2':
            try:
                num = int(input("Enter an integer to check for primality: "))
                result = is_prime(num)
                if isinstance(result, bool):
                    print(f"{num} is {'a prime number' if result else 'not a prime number'}.")
                else:
                    print(result)
            except ValueError:
                print("Invalid input. Please enter an integer.")

        elif choice == '3':
            try:
                num = int(input("Enter a non-negative integer to check for Armstrong: "))
                result = is_armstrong(num)
                if isinstance(result, bool):
                    print(f"{num} is {'an Armstrong number' if result else 'not an Armstrong number'}.")
                else:
                    print(result)
            except ValueError:
                print("Invalid input. Please enter an integer.")

        elif choice == '4':
            print("Exiting program. Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number between 1 and 4.")

# Run the main menu
main_menu()
```
