```python
# Q38. Generate random temperature data using NumPy and plot a histogram

import numpy as np
import matplotlib.pyplot as plt

# 1. Generate random temperature data using NumPy
#    Let's assume temperatures follow a normal distribution with:
#    - mean (loc): 25 degrees Celsius
#    - standard deviation (scale): 5 degrees Celsius
#    - size: 1000 data points
#    We'll also clip the temperatures to a reasonable range (e.g., 10 to 40) to prevent extreme values.
np.random.seed(42) # for reproducibility
temperatures = np.random.normal(loc=25, scale=5, size=1000)

# Clip values to a realistic range, for example, between 10 and 40 degrees Celsius
temperatures = np.clip(temperatures, 10, 40)

print(f"Generated {len(temperatures)} temperature data points. First 10 samples: {temperatures[:10].round(2)}")
print(f"Mean temperature: {np.mean(temperatures):.2f}°C")
print(f"Standard deviation: {np.std(temperatures):.2f}°C")

# 2. Plot a histogram representing the temperature distribution
plt.figure(figsize=(10, 6)) # Set the figure size for better readability

# Create the histogram
#   - bins: Number of bins to divide the data into (e.g., 30 bins)
#   - edgecolor: Color of the bin edges for better visual separation
#   - alpha: Transparency of the bars (0.0 to 1.0)
plt.hist(temperatures, bins=30, edgecolor='black', alpha=0.7, color='skyblue')

# 3. Add titles and labels for clarity
plt.title('Distribution of Random Temperature Data', fontsize=16) # Title of the histogram
plt.xlabel('Temperature (°C)', fontsize=12) # Label for the x-axis
plt.ylabel('Frequency', fontsize=12) # Label for the y-axis
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add a grid for easier reading of frequencies

# 4. Display the plot
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show() # Show the generated plot
```
