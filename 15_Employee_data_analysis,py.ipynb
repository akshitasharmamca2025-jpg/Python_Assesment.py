```python
# =========================
# Q15 Employee Data Analysis with Pandas
# =========================

# Import the pandas library for data manipulation
import pandas as pd

# 1. Create a dummy CSV file for demonstration purposes
# In a real scenario, you would already have an 'employees.csv' file.
csv_data = """
EmployeeID,Name,Department,Salary
1,Alice,HR,60000
2,Bob,IT,75000
3,Charlie,HR,65000
4,David,Finance,80000
5,Eve,IT,90000
6,Frank,HR,55000
7,Grace,Finance,82000
8,Heidi,IT,70000
"""

# Write the data to a CSV file
with open('employees.csv', 'w') as f:
    f.write(csv_data)

print("Created 'employees.csv' with sample employee data.")

# 2. Read employee data from the CSV file into a Pandas DataFrame
try:
    df = pd.read_csv('employees.csv')
    print("\nEmployee Data Loaded:")
    print(df.head())
except FileNotFoundError:
    print("Error: 'employees.csv' not found. Please ensure the file exists.")
    exit()
except Exception as e:
    print(f"An error occurred while reading the CSV: {e}")
    exit()

# 3. Display department-wise average salary
# Group the DataFrame by 'Department' and calculate the mean of 'Salary'
department_avg_salary = df.groupby('Department')['Salary'].mean()
print("\nDepartment-wise Average Salary:")
print(department_avg_salary)

# 4. Find the highest salary employee in each department
# Group by 'Department' and then use idxmax() to find the index of the max salary in each group
# Then use .loc to get the full row for that index
highest_salary_employees = df.loc[df.groupby('Department')['Salary'].idxmax()]
print("\nEmployee with Highest Salary per Department:")
print(highest_salary_employees[['Department', 'Name', 'Salary']])
```
