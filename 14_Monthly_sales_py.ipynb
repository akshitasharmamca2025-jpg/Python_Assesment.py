```python
# =========================
# Q14 Monthly Sales Line Chart
# =========================

# Import the necessary library for plotting
import matplotlib.pyplot as plt

# Import numpy for numerical operations, specifically for creating sample data
import numpy as np

# 1. Prepare the data
# Define the months (x-axis data)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
# Define sales data for Company A (y-axis data)
sales_company_a = [150, 180, 220, 200, 250, 230, 280, 300, 270, 320, 350, 380]
# Define sales data for Company B to demonstrate legend
sales_company_b = [120, 140, 190, 170, 210, 200, 240, 260, 250, 290, 310, 340]

# 2. Create the line chart
plt.figure(figsize=(10, 6)) # Set the figure size for better readability

# Plot sales data for Company A
plt.plot(months, sales_company_a, marker='o', linestyle='-', color='b', label='Company A Sales')
# Plot sales data for Company B
plt.plot(months, sales_company_b, marker='x', linestyle='--', color='r', label='Company B Sales')

# 3. Add visualization enhancements
# Add a title to the chart
plt.title('Monthly Sales Data for Companies A and B', fontsize=16)
# Add label for the x-axis
plt.xlabel('Month', fontsize=12)
# Add label for the y-axis
plt.ylabel('Sales (in thousands $)', fontsize=12)
# Add a legend to identify the plotted lines
plt.legend(fontsize=10)
# Add a grid for better readability of data points
plt.grid(True, linestyle='--', alpha=0.7)

# 4. Display the chart
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show() # Show the plot
```
