```python
# =========================
# Q21 Student Marks Analysis with NumPy and Pandas
# =========================

import numpy as np
import pandas as pd

# 1. Generate a NumPy array of student marks
# Let's assume 5 students and 3 subjects (Math, Science, English)
# Marks can range from 40 to 100
num_students = 5
num_subjects = 3

# Generate random integer marks for demonstration
marks_array = np.random.randint(40, 101, size=(num_students, num_subjects))

print("NumPy Array of Student Marks:\n", marks_array)

# 2. Convert the NumPy array into a Pandas DataFrame
subject_names = ['Math', 'Science', 'English']
student_names = [f'Student_{i+1}' for i in range(num_students)]

marks_df = pd.DataFrame(marks_array, columns=subject_names, index=student_names)

print("\nPandas DataFrame of Student Marks:\n", marks_df)

# 3. Display highest marks (overall)
highest_overall_mark = marks_df.values.max()
print(f"\nHighest Mark (Overall): {highest_overall_mark}")

# 4. Display average marks (overall)
average_overall_mark = marks_df.values.mean()
print(f"Average Mark (Overall): {average_overall_mark:.2f}")

# 5. Display subject-wise statistics
print("\nSubject-wise Statistics:")
subject_stats = marks_df.describe().loc[['mean', 'max', 'min']]
print(subject_stats)

# Optionally, also display highest and average marks per student
print("\nStudent-wise Highest Marks:")
print(marks_df.max(axis=1))

print("\nStudent-wise Average Marks:")
print(marks_df.mean(axis=1))
```
