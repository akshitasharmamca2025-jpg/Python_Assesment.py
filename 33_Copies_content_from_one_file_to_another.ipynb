```python
# Q33. Create a program that copies contents from one file to another and displays the total number of words copied.

def copy_file_and_count_words():
    """
    Copies content from a source file to a destination file
    and counts the total number of words copied.
    """
    source_file_name = input("Enter the name of the source file: ")
    destination_file_name = input("Enter the name of the destination file: ")

    total_words_copied = 0

    try:
        # Open the source file in read mode
        with open(source_file_name, 'r', encoding='utf-8') as source_file:
            content = source_file.read()

            # Count words in the content
            words = content.split()
            total_words_copied = len(words)

        # Open the destination file in write mode and write the content
        with open(destination_file_name, 'w', encoding='utf-8') as destination_file:
            destination_file.write(content)

        print(f"Content successfully copied from '{source_file_name}' to '{destination_file_name}'.")
        print(f"Total words copied: {total_words_copied}")

    except FileNotFoundError:
        print(f"Error: The file '{source_file_name}' was not found. Please ensure it exists.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Run the program
copy_file_and_count_words()

# Example usage:
# To test this, you can first create a dummy source file like this:
# with open('source.txt', 'w', encoding='utf-8') as f:
#     f.write('This is a test file.\n')
#     f.write('It has multiple lines and words.\n')
#     f.write('The quick brown fox jumps over the lazy dog.')
# print("Created 'source.txt' for testing.")

# Then, run this cell and enter 'source.txt' as the source and 'destination.txt' as the destination.
```
