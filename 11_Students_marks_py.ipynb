```python
#11.Students Marks File:
# 1. Create a text file containing student marks
file_name = 'student_marks.txt'
student_data = [
    "Alice,85",
    "Bob,92",
    "Charlie,78",
    "David,65",
    "Eve,95",
    "Frank,88",
    "Grace,72",
    "Heidi,80"
]

with open(file_name, 'w') as f:
    for line in student_data:
        f.write(line + '\n')

print(f"Created '{file_name}' with sample student marks.")

# 2. Read the file and process the data
students = []
total_marks = 0

try:
    with open(file_name, 'r') as f:
        for line in f:
            name, mark_str = line.strip().split(',')
            mark = int(mark_str)
            students.append({'name': name, 'mark': mark})
            total_marks += mark
except FileNotFoundError:
    print(f"Error: The file '{file_name}' was not found.")
except ValueError:
    print("Error: Invalid data format in the file. Marks must be integers.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")

if students:
    # 3. Display the topper
    topper = max(students, key=lambda s: s['mark'])
    print(f"\nTopper: {topper['name']} with {topper['mark']} marks")

    # 4. Display average marks
    average_marks = total_marks / len(students)
    print(f"Average Marks: {average_marks:.2f}")

    # 5. Display students scoring below average
    below_average_students = [s for s in students if s['mark'] < average_marks]
    print("\nStudents scoring below average:")
    if below_average_students:
        for student in below_average_students:
            print(f"- {student['name']} ({student['mark']} marks)")
    else:
        print("No students scored below average.")
else:
    print("No student data was read from the file.")
```
