```python
# =======================================================
# Q16 Generate Even Fibonacci Numbers up to N
# =======================================================

def generate_even_fibonacci(N):
    """
    Generates Fibonacci numbers up to N and returns a list of only the even ones.

    Args:
        N (int): The maximum value for Fibonacci numbers to be considered.

    Returns:
        list: A list containing even Fibonacci numbers up to N.
    """
    # Initialize the first two Fibonacci numbers
    a, b = 0, 1
    even_fib_numbers = []

    # Loop to generate Fibonacci numbers until 'a' exceeds N
    while a <= N:
        # Check if the current Fibonacci number 'a' is even
        if a % 2 == 0:
            even_fib_numbers.append(a)

        # Generate the next Fibonacci number
        a, b = b, a + b

    return even_fib_numbers

# Example usage:
N_limit = 100
even_fib_list = generate_even_fibonacci(N_limit)

print(f"Even Fibonacci numbers up to {N_limit}: {even_fib_list}")

N_limit_2 = 500
even_fib_list_2 = generate_even_fibonacci(N_limit_2)
print(f"Even Fibonacci numbers up to {N_limit_2}: {even_fib_list_2}")
```
