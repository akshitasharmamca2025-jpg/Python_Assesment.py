```python
# =========================
# Q6 Merge Lists
# =========================

# Creating two lists
list1 = [2, 4, 6, 9]
list2 = [3, 6, 9, 12]

# Merge lists and remove duplicates
merged = list(set(list1 + list2))

# Sort list in descending order
merged.sort(reverse=True)

# Display merged list
print("Merged List:", merged)

# Check numbers divisible by 3 and 5
print("Numbers divisible by 3 and 5:")

for i in merged:

    if i % 3 == 0 and i % 5 == 0:
        print(i)
```
