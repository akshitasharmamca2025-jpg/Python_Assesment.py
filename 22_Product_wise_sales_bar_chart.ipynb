```python
# =========================
# Q22 Product-wise Sales Bar Chart
# =========================

import pandas as pd
import matplotlib.pyplot as plt
import os

# 1. Create a dummy CSV file for demonstration
# In a real scenario, you would have an existing 'sales_data.csv' file.
csv_content = """
Product,Sales
Laptop,1200
Mouse,300
Keyboard,500
Laptop,800
Monitor,1500
Mouse,250
Keyboard,600
Monitor,1000
"""

file_name = 'sales_data.csv'

# Write the content to the CSV file
with open(file_name, 'w') as f:
    f.write(csv_content)

print(f"Created '{file_name}' with sample sales data.")

# 2. Read the sales data from the CSV file using Pandas
try:
    df_sales = pd.read_csv(file_name)
    print("\nSales Data Loaded:")
    print(df_sales.head())
except FileNotFoundError:
    print(f"Error: The file '{file_name}' was not found. Please ensure the file exists.")
    exit()
except Exception as e:
    print(f"An error occurred while reading the CSV: {e}")
    exit()

# 3. Calculate product-wise total sales
product_sales = df_sales.groupby('Product')['Sales'].sum().sort_values(ascending=False)
print("\nProduct-wise Total Sales:")
print(product_sales)

# 4. Plot a bar chart representing product-wise sales using Matplotlib
plt.figure(figsize=(10, 6))
product_sales.plot(kind='bar', color='skyblue')

# Add chart enhancements
plt.title('Product-wise Sales', fontsize=16)
plt.xlabel('Product', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Adjust layout to prevent labels from overlapping

# Display the chart
plt.show()

# Optional: Clean up the dummy CSV file after use
# os.remove(file_name)
# print(f"\nRemoved '{file_name}'.")
```
