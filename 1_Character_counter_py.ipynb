```python
# Q1. Character Counter and Frequency Analyzer

def analyze_paragraph(paragraph):
    """
    Analyzes a given paragraph to count uppercase letters, lowercase letters,
    digits, spaces, and special characters.
    Displays the counts in descending order based on frequency.
    """

    # Initialize counters for each character type
    uppercase_count = 0
    lowercase_count = 0
    digit_count = 0
    space_count = 0
    special_char_count = 0

    # Iterate through each character in the paragraph
    for char in paragraph:
        if char.isupper():
            uppercase_count += 1
        elif char.islower():
            lowercase_count += 1
        elif char.isdigit():
            digit_count += 1
        elif char.isspace():
            space_count += 1
        else:
            special_char_count += 1

    # Store the counts in a dictionary for easy sorting
    character_counts = {
        "Uppercase Letters": uppercase_count,
        "Lowercase Letters": lowercase_count,
        "Digits": digit_count,
        "Spaces": space_count,
        "Special Characters": special_char_count
    }

    # Sort the dictionary items by their counts in descending order
    # `item[1]` refers to the count (value) in each key-value pair
    sorted_counts = sorted(character_counts.items(), key=lambda item: item[1], reverse=True)

    # Display the results
    print("\n--- Character Analysis Results (Sorted by Frequency) ---")
    for char_type, count in sorted_counts:
        print(f"{char_type}: {count}")
    print("----------------------------------------------------------")

# Main part of the program:
# 1. Accept a paragraph from the user
user_paragraph = input("Enter a paragraph: ")

# 2. Call the analysis function
analyze_paragraph(user_paragraph)
```
