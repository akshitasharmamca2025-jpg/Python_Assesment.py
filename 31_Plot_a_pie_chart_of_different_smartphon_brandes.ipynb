```python
# Q31. Plot a pie chart showing market share of different smartphone brands
# and highlight the brand with maximum market share also add comments

import matplotlib.pyplot as plt
import numpy as np

# 1. Sample Data for Smartphone Brands and Market Share
#    Brands represent the labels for each slice of the pie chart.
#    Market share represents the size of each slice.
brands = ['Samsung', 'Apple', 'Xiaomi', 'Huawei', 'Oppo', 'Others']
market_share = [22, 18, 12, 10, 8, 30] # Percentages, should sum up to 100 or be relative proportions

# 2. Find the brand with the maximum market share to highlight it
#    np.argmax returns the index of the maximum value in the array.
max_share_index = np.argmax(market_share)

# 3. Create an 'explode' tuple to highlight a slice
#    'explode' is a tuple of the same length as the data, where each value
#    indicates how far to 'explode' a slice from the center (e.g., 0.1 for 10% of the radius).
#    We create a tuple of zeros and set the value at max_share_index to 0.1.
explode = [0] * len(brands)
explode[max_share_index] = 0.1 # Highlight the brand with maximum market share

# 4. Define colors for the pie slices (optional, but good for aesthetics)
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#CC99FF', '#C0C0C0']

# 5. Create the pie chart
plt.figure(figsize=(9, 9)) # Set the figure size for a larger, clearer chart

# plt.pie() creates the pie chart.
#   - market_share: The data that defines the size of each wedge.
#   - explode: Determines which wedges are offset from the center.
#   - labels: Text labels for each wedge.
#   - colors: Custom colors for each wedge.
#   - autopct: Formats the percentage value displayed on each wedge. '%1.1f%%' means one decimal place.
#   - shadow: Adds a shadow effect to the pie chart.
#   - startangle: Rotates the start of the pie chart by a given angle (e.g., 90 degrees starts the first slice from the top).
plt.pie(market_share, explode=explode, labels=brands, colors=colors,
        autopct='%1.1f%%', shadow=True, startangle=140)

# 6. Add a title to the pie chart for clarity
plt.title('Smartphone Market Share by Brand', fontsize=16)

# 7. Ensure the circle is drawn proportionally (aspect ratio)
#    'equal' ensures that pie is drawn as a circle rather than an ellipse.
plt.axis('equal')

# 8. Add comments for the highlighted brand
#    This text annotation specifies which brand has the maximum market share.
max_brand = brands[max_share_index]
max_share_value = market_share[max_share_index]
plt.text(1.2, 1.0, f'Note: {max_brand} has the highest market share ({max_share_value}%)',
         fontsize=12, bbox=dict(facecolor='yellow', alpha=0.5))

# 9. Display the chart
plt.show()
```
