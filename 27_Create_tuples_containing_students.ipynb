```python
# Q27. Create tuples containing student names and courses. Convert them into sets to identify students enrolled in multiple courses

# 1. Create tuples containing student names and courses
student_courses_data = [
    ("Alice", "Math"),
    ("Bob", "Physics"),
    ("Alice", "Chemistry"),
    ("Charlie", "Math"),
    ("Bob", "Chemistry"),
    ("Alice", "Physics"),
    ("David", "Biology"),
    ("Charlie", "Physics")
]

print("Original student-course data (list of tuples):")
print(student_courses_data)

# 2. Convert the list of tuples into a set for efficient lookup and to remove accidental duplicate entries
student_courses_set = set(student_courses_data)

print("\nStudent-course enrollments (set of tuples):")
print(student_courses_set)

# 3. Identify students enrolled in multiple courses
# We can use a dictionary to map each student to the set of courses they are taking.
student_enrollments = {}
for student, course in student_courses_set:
    if student not in student_enrollments:
        student_enrollments[student] = set()
    student_enrollments[student].add(course)

print("\nStudent enrollments (student: {courses}):")
for student, courses in student_enrollments.items():
    print(f"{student}: {courses}")

# Filter for students who are enrolled in more than one course
students_in_multiple_courses = []
for student, courses in student_enrollments.items():
    if len(courses) > 1:
        students_in_multiple_courses.append(student)

print("\nStudents enrolled in multiple courses:")
if students_in_multiple_courses:
    print(students_in_multiple_courses)
else:
    print("No students are enrolled in multiple courses.")
```
