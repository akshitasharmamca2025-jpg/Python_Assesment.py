```python
# ==================================================
# Q25 Word Frequencies from File and Top Five Words
# ==================================================

import re # For regular expression operations (cleaning text)
from collections import Counter # For efficient word counting
import os # For file operations (creating and potentially removing dummy file)

def analyze_word_frequencies(filename):
    """
    Reads a text file, counts word frequencies, and displays the top 5 most frequent words.
    """
    word_counts = Counter()
    total_words = 0

    try:
        with open(filename, 'r', encoding='utf-8') as f:
            for line in f:
                # Convert to lowercase and find all alphabetic sequences (words)
                words = re.findall(r'\b[a-z]+\b', line.lower())
                word_counts.update(words)
                total_words += len(words)

        if not word_counts:
            print(f"No words found in the file '{filename}'.")
            return

        print(f"\n--- Word Frequency Analysis for '{filename}' ---")
        print(f"Total unique words: {len(word_counts)}")
        print(f"Total words (including repetitions): {total_words}")

        # Display the top five most frequently used words
        print("\nTop 5 Most Frequently Used Words:")
        for word, count in word_counts.most_common(5):
            print(f"'{word}': {count} times")

    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found. Please ensure the file exists and the path is correct.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# --- Example Usage ---
# 1. Create a dummy text file for testing
dummy_file_name = 'sample_text_for_q25.txt'
dummy_content = """
This is a sample text file. This file contains sample words.
Word frequency analysis is interesting. This is a good test.
Test for words. Words, words, words!
"""

with open(dummy_file_name, 'w', encoding='utf-8') as f:
    f.write(dummy_content)

print(f"Created dummy file: '{dummy_file_name}'")

# 2. Analyze word frequencies from the dummy file
analyze_word_frequencies(dummy_file_name)

# Optional: Clean up the dummy file after use
# os.remove(dummy_file_name)
# print(f"\nRemoved '{dummy_file_name}'.")
```
