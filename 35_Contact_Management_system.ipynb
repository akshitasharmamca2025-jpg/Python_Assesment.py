```python
# Q35. Contact Management System

# Using a dictionary to store contacts. The key will be the contact's name,
# and the value will be another dictionary containing phone and email.
contacts = {} # Initialize an empty dictionary to store contacts

def add_contact():
    """
    Adds a new contact to the system.
    Prompts the user for contact's name, phone, and email.
    Handles cases where the name is empty or already exists.
    """
    name = input("Enter contact name: ").strip().title() # Get name, remove whitespace, and title case it
    if not name:
        print("Contact name cannot be empty.")
        return # Exit if name is empty
    if name in contacts:
        print(f"Contact '{name}' already exists. Use update option to change details.")
    else:
        phone = input("Enter phone number: ").strip() # Get phone number
        email = input("Enter email address: ").strip() # Get email address
        contacts[name] = {"phone": phone, "email": email} # Store contact details in the dictionary
        print(f"Contact '{name}' added successfully.")

def update_contact():
    """
    Updates an existing contact's phone and/or email.
    Prompts the user for the contact's name and new details.
    Allows leaving fields blank to keep current values.
    """
    name = input("Enter name of contact to update: ").strip().title()
    if not name:
        print("Contact name cannot be empty.")
        return
    if name in contacts:
        print(f"Current details for '{name}':")
        print(f"  Phone: {contacts[name]['phone']}")
        print(f"  Email: {contacts[name]['email']}")

        new_phone = input("Enter new phone number (leave blank to keep current): ").strip()
        new_email = input("Enter new email address (leave blank to keep current): ").strip()

        if new_phone: # Update phone if new_phone is not empty
            contacts[name]["phone"] = new_phone
        if new_email: # Update email if new_email is not empty
            contacts[name]["email"] = new_email
        print(f"Contact '{name}' updated successfully.")
    else:
        print(f"Contact '{name}' not found.")

def search_contact():
    """
    Searches for a contact by name and displays its details.
    """
    name = input("Enter name of contact to search: ").strip().title()
    if not name:
        print("Contact name cannot be empty.")
        return
    if name in contacts:
        print(f"\n--- Contact Details for '{name}' ---")
        print(f"  Phone: {contacts[name]['phone']}")
        print(f"  Email: {contacts[name]['email']}")
        print("------------------------------------")
    else:
        print(f"Contact '{name}' not found.")

def delete_contact():
    """
    Deletes a contact from the system.
    """
    name = input("Enter name of contact to delete: ").strip().title()
    if not name:
        print("Contact name cannot be empty.")
        return
    if name in contacts:
        del contacts[name] # Delete the contact entry
        print(f"Contact '{name}' deleted successfully.")
    else:
        print(f"Contact '{name}' not found.")

def view_all_contacts():
    """
    Displays all contacts in the system in alphabetical order by name.
    """
    if not contacts:
        print("No contacts in the system.")
        return

    print("\n--- All Contacts ---")
    for name, details in sorted(contacts.items()): # Iterate through sorted contacts
        print(f"Name: {name}")
        print(f"  Phone: {details['phone']}")
        print(f"  Email: {details['email']}")
        print("--------------------")
    print("--------------------")

def main():
    """
    Main function to run the contact management system menu.
    Continuously displays options until the user chooses to exit.
    """
    while True:
        print("\n--- Contact Management System ---")
        print("1. Add Contact")
        print("2. Update Contact")
        print("3. Search Contact")
        print("4. Delete Contact")
        print("5. View All Contacts")
        print("6. Exit")
        print("---------------------------------")

        choice = input("Enter your choice (1-6): ").strip()

        if choice == '1':
            add_contact()
        elif choice == '2':
            update_contact()
        elif choice == '3':
            search_contact()
        elif choice == '4':
            delete_contact()
        elif choice == '5':
            view_all_contacts()
        elif choice == '6':
            print("Exiting Contact Management System. Goodbye!")
            break # Exit the while loop to end the program
        else:
            print("Invalid choice. Please enter a number between 1 and 6.")

# Ensures that the main() function runs only when the script is executed directly
if __name__ == "__main__":
    main()
```
